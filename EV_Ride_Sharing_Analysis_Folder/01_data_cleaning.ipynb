```python
import os
import pandas as pd
import pyodbc

# --- Connection details (you already confirmed these) ---
server = r'LAPTOP-IG80BF3U\ABSERVERPBI'
database = 'EV_Ride_Sharing'

conn_str = (
    "DRIVER={ODBC Driver 17 for SQL Server};"
    f"SERVER={server};DATABASE={database};"
    "Trusted_Connection=yes;Encrypt=no"
)

# Connect to SQL Server
cnxn = pyodbc.connect(conn_str)
cursor = cnxn.cursor()
cursor.fast_executemany = True
print("✅ SQL Server connection established.")

# --- Folder containing your cleaned CSV files ---
DATA_DIR = r"C:\Users\style\EV_Ride_Sharing_Full_Dataset\data_processed"

# Map CSV filenames → SQL table names
table_map = {
    'drivers.csv': 'drivers',
    'vehicles.csv': 'vehicles',
    'customers.csv': 'customers',
    'payments.csv': 'payments',
    'charging_stations.csv': 'charging_stations',
    'rides.csv': 'rides'
}


# --- Load each CSV into its corresponding SQL table ---
for fname, table in table_map.items():
    path = os.path.join(DATA_DIR, fname)
    
    if not os.path.exists(path):
        print(f"⚠️ Skipping {fname} — file not found.")
        continue

    print(f"\n📦 Loading {fname} → [{table}] ...")

    # Read CSV into DataFrame
    df = pd.read_csv(path)
    rows = len(df)

    # Build SQL insert statement dynamically
    cols = list(df.columns)
    col_names = ",".join([f"[{c}]" for c in cols])
    placeholders = ",".join(["?"] * len(cols))
    sql = f"INSERT INTO dbo.{table} ({col_names}) VALUES ({placeholders})"

    try:
        cursor.executemany(sql, df.itertuples(index=False, name=None))
        cnxn.commit()
        print(f"✅ Loaded {rows:,} rows into [{table}] successfully.")
    except Exception as e:
        print(f"❌ Error loading [{table}]: {e}")
        cnxn.rollback()

cursor.close()
cnxn.close()
print("\n🎉 All CSVs processed — data successfully loaded into SQL Server!")
```
